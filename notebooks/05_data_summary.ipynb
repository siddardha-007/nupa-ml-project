```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/customers.csv")

df["Age"] = df["Age"].fillna(df["Age"].median())

df["AnnualIncome"] = df["AnnualIncome"].fillna(
    df["AnnualIncome"].median()
)

print(df.head())
```

```python
city_spending = df.groupby("City")["SpendingScore"].mean()

city_spending
```

```python
city_spending = (
    df.groupby("City", as_index=False)["SpendingScore"]
      .mean()
)

print(city_spending)
```

```python
summary = df.groupby("City").agg(
    Customer_Count=("CustomerID", "count"),
    Average_Age=("Age", "mean"),
    Average_Income=("AnnualIncome", "mean"),
    Average_Spending=("SpendingScore", "mean"),
    Maximum_Spending=("SpendingScore", "max")
)

summary
```

```python
result = df.groupby(
    ["City", "Membership"],
    as_index=False
).agg(
    Customers=("CustomerID", "count"),
    Average_Income=("AnnualIncome", "mean"),
    Average_Spending=("SpendingScore", "mean")
)

print(result.round(2))
```

```python
pivot = pd.pivot_table(
    df,
    values="SpendingScore",
    index="City",
    columns="Membership",
    aggfunc="mean"
)

print(pivot.round(2))
```

```python
counts = pd.crosstab(
    df["City"],
    df["Membership"]
)

print(counts)
```

```python
city_summary = df.groupby(
    "City",
    as_index=False
).agg(
    Average_Spending=("SpendingScore", "mean"),
    Customer_Count=("CustomerID", "count")
)


city_summary = city_summary.sort_values(
    "Average_Spending",
    ascending=False
)

print(city_summary)
```

```python
df["City_Average_Income"] = (
    df.groupby("City")["AnnualIncome"]
      .transform("mean")
)

df["Income_vs_City_Average"] = (
    df["AnnualIncome"] - df["City_Average_Income"]
)

print(
    df[
        [
            "CustomerID",
            "City",
            "AnnualIncome",
            "City_Average_Income",
            "Income_vs_City_Average"
        ]
    ]
)
df
```

```python
import pandas as pd
import numpy as np

customers = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105],
    "Name": ["Asha", "Rahul", "Priya", "Kiran", "Arjun"],
    "City": ["Hyderabad", "Bengaluru", "Chennai", "Mumbai", "Hyderabad"]
})

transactions = pd.DataFrame({
    "TransactionID": [1, 2, 3, 4, 5, 6],
    "CustomerID": [101, 102, 101, 103, 106, 105],
    "Amount": [1200, 2500, 800, 3200, 1500, 4000]
})

print(customers)
print(transactions)
```

```python
inner = pd.merge(
    customers,
    transactions,
    on="CustomerID",
    how="inner"
)

print(inner)
```

```python
left = pd.merge(
    customers,
    transactions,
    on="CustomerID",
    how="left"
)

print(left)
```

```python
right = pd.merge(
    customers,
    transactions,
    on="CustomerID",
    how="right"
)

print(right)
```

```python
outer = pd.merge(
    customers,
    transactions,
    on="CustomerID",
    how="outer",
    indicator=True
)

print(outer)
```

```python
print(customers["CustomerID"].is_unique)
print(transactions["CustomerID"].is_unique)
```

```python
result = customers.merge(
    transactions,
    on="CustomerID",
    how="left",
    validate="one_to_many"
)

print(result)
```

```python
jan = pd.DataFrame({
    "CustomerID": [101, 102],
    "Amount": [1200, 2500]
})

feb = pd.DataFrame({
    "CustomerID": [103, 104],
    "Amount": [1800, 3200]
})

combined = pd.concat([jan, feb], ignore_index=True)

print(combined)
```

```python
names = pd.DataFrame({
    "Name": ["Asha", "Rahul", "Priya"]
})

scores = pd.DataFrame({
    "Score": [85, 92, 78]
})

combined = pd.concat([names, scores], axis=1)

print(combined)
```

```python
customer_indexed = customers.set_index("CustomerID")
transactions_indexed = transactions.set_index("CustomerID")

result = customer_indexed.join(
    transactions_indexed,
    how="left"
)

print(result)
```

```python
# Attach customer details to every transaction
purchase_data = transactions.merge(
    customers,
    on="CustomerID",
    how="left",
    validate="many_to_one",
    indicator=True
)

print(purchase_data)
```

```python
unmatched = purchase_data[
    purchase_data["_merge"] == "left_only"
]

print(unmatched)
```

```python
customer_summary = (
    purchase_data[purchase_data["_merge"] == "both"]
    .groupby(["CustomerID", "Name", "City"], as_index=False)
    .agg(
        Total_Spending=("Amount", "sum"),
        Number_Of_Transactions=("TransactionID", "count"),
        Average_Purchase=("Amount", "mean")
    )
)

(customer_summary.sort_values("Total_Spending", ascending=False))
```
