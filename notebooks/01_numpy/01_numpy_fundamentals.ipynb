```python
import numpy as np

marks = np.array([
    [78, 85, 92],
    [67, 88, 76],
    [90, 95, 89],
    [35, 42, 38],
    [55, 72, 68]
])

# 1. Calculate average marks per subject
subject_means = marks.mean(axis=0)
print("Subject averages:", subject_means)

# 2. Calculate each student's average
student_means = marks.mean(axis=1)
print("Student averages:", student_means)

# 3. Find students with an average >= 60
selected_students = student_means >= 60
print("Selected averages:", student_means[selected_students])

# 4. Rank students by average marks
ranking = np.argsort(student_means)[::-1]
print("Ranking (student indexes):", ranking)

# 5. Find all individual marks below 40
low_marks = marks[marks < 40]
print("Marks below 40:", low_marks)
```

```python
hours = np.array([1, 2, 3, 4, 5], dtype=float)

scores = np.array([35, 40, 50, 55, 65], dtype=float)

# Calculate slope and intercept using least squares
X = np.column_stack([hours, np.ones_like(hours)])

weights, residuals, rank, singular_values = np.linalg.lstsq(
    X, scores, rcond=None
)

slope, intercept = weights

predictions = slope * hours + intercept

print("Slope:", slope)
print("Intercept:", intercept)
print("Predictions:", predictions)
print("Actual:", scores)
```
