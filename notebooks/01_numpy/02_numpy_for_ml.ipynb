```python
data = np.array([10, 20, np.nan, 40, 50])

print(np.isnan(data))
print(np.nanmean(data))    # 30.0
print(np.nanmedian(data))  # 30.0
```

```python
data = np.array([10.0, np.nan, 30.0, np.nan, 50.0])

missing_mask = np.isnan(data)

print("Missing count:", missing_mask.sum())

mean_value = np.nanmean(data)

cleaned = np.where(missing_mask, mean_value, data)

print(cleaned)
# [10. 30. 30. 30. 50.]
```

```python
import numpy as np

# Rows = students; columns = features
X = np.array([
    [18, 75, 80],
    [20, 85, 90],
    [19, np.nan, 70],
    [21, 65, 60],
    [22, 95, 95]
], dtype=np.float64)

# 1. Inspect the dataset
print("Shape:", X.shape)
print("Data type:", X.dtype)

# 2. Count missing values per feature
print("Missing values:", np.isnan(X).sum(axis=0))

# 3. Impute missing values with each column's mean
column_means = np.nanmean(X, axis=0)
rows, cols = np.where(np.isnan(X))
X[rows, cols] = column_means[cols]

# 4. Standardize each feature
means = X.mean(axis=0)
stds = X.std(axis=0)

# Guard against features with zero variance
stds[stds == 0] = 1

X_scaled = (X - means) / stds

print("Prepared dataset:")
print(X_scaled)
```
