```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

data = load_breast_cancer()

X = data.data
y = data.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.15,
    random_state=42,
    stratify=y
)
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.1765,
    random_state=42,
    stratify=y_train
)
```

```python
print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

```python
import numpy as np

print("Train:", np.bincount(y_train) / len(y_train))
print("Validation:", np.bincount(y_val) / len(y_val))
print("Test:", np.bincount(y_test) / len(y_test))
```

## Train / Validation / Test Summary

Machine learning datasets can be divided into training, validation, and
test sets.

The training set is used to learn model parameters.

The validation set is used for model comparison, hyperparameter tuning,
and decision making during development.

The test set is kept separate and is used only for the final evaluation
of the selected model.

Keeping the test set untouched helps provide a more reliable estimate of
how the model will perform on unseen data.
