```python
def inspect_dataset(df):
    print("=" * 45)
    print("DATASET INSPECTION REPORT")
    print("=" * 45)

    print(f"\nRows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")

    print("\nColumn data types:")
    print(df.dtypes)

    print("\nMissing values:")
    print(df.isnull().sum())

    print("\nMissing percentages:")
    print((df.isnull().mean() * 100).round(2))

    print("\nDuplicate rows:", df.duplicated().sum())

    print("\nNumerical summary:")
    print(df.describe())

    print("\nCategorical summary:")
    print(df.describe(include="object"))


inspect_dataset(df)
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/customers.csv")

print(df.head())
```

```python
# Select one column — returns a Series
ages = df["Age"]

# Select multiple columns — returns a DataFrame
features = df[["Age", "AnnualIncome", "SpendingScore"]]

print(type(ages))
print(type(features))
```

```python
X = df[["Age", "AnnualIncome", "SpendingScore"]]

print(X.shape)
```

```python
# First row by position
print(df.iloc[0])

# First three rows
print(df.iloc[:3])

# Rows 0–2, columns Age and City
print(df.loc[0:2, ["Age", "City"]])

# Second row, third column by position
print(df.iloc[1, 2])
```

```python
indexed_df = df.set_index("CustomerID")

print(indexed_df.loc[101])
print(indexed_df.loc[[101, 105, 109]])
indexed_df
```

```python
older_customers = df[df["Age"] > 30]

print(older_customers)
```

```python
df[df["AnnualIncome"] >= 800000]
df[df["SpendingScore"] < 50]
df[df["City"] == "Hyderabad"]
df[df["Membership"] == "Gold"]
```

```python
# Customers older than 25 AND spending score above 60
result = df[
    (df["Age"] > 25) &
    (df["SpendingScore"] > 60)
]

print(result)
```

```python
cities = ["Hyderabad", "Bengaluru"]

result = df[df["City"].isin(cities)]

print(result)
```

```python
missing_age = df[df["Age"].isna()]

known_age = df[df["Age"].notna()]

display(
    known_age,
    missing_age
)
```

```python
# Highest spending score first
print(df.sort_values("SpendingScore", ascending=False))

# Lowest income first
print(df.sort_values("AnnualIncome", ascending=True))
```

```python
result = df.sort_values(
    by=["City", "AnnualIncome"],
    ascending=[True, False]
)

print(result)
```

```python
top_customers = df.nlargest(3, "SpendingScore")

print(top_customers)
```

```python
# Check membership in a list
print(df[df["City"].isin(["Mumbai", "Chennai"])])

# Search text
print(df[df["City"].str.contains("pur", case=False, na=False)])

# Filter using query syntax
print(df.query("Age > 25 and SpendingScore >= 60"))
```

```python
good_customers = df[
    (df["AnnualIncome"]>=500000) &
    (df["SpendingScore"]>=60) &
    (df["Age"].notna())
]

good_customers = good_customers.sort_values(
    "SpendingScore" ,
    ascending = False
)

good_customers
```
